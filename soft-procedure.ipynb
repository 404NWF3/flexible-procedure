# 多品种小批量柔性生产线调度优化（生产制造·柔性制造）



## 一、案例背景

某精密制造企业拥有一条由**五个加工中心**组成的柔性生产线，可生产**四种不同类型**的产品。仿真时段为**一周（5 个工作日，每天 8 小时）**。

## 二、产品与工艺路线

| 产品 | 占比 | 工艺路线（各工位加工时间，分钟） |
|------|------|------|
| P1 | 30% | M1(5) → M2(8) → M3(6) → M4(4) → M5(3) |
| P2 | 25% | M1(4) → M3(7) → M4(5) → M5(6) |
| P3 | 25% | M2(6) → M3(5) → M5(8) |
| P4 | 20% | M1(3) → M4(9) → M5(4) |

> 各工位加工时间服从**正态分布**，标准差为均值的 10%。

## 三、工位配置

| 工位 | 机床数量 | 可加工产品 |
|------|----------|------------|
| M1 | 2 台 | P1, P2, P4 |
| M2 | 1 台 | P1, P3 |
| M3 | 2 台 | P1, P2, P3 |
| M4 | 1 台 | P1, P2, P4 |
| M5 | 2 台 | P1, P2, P3, P4 |

## 四、生产计划

- 产品毛坯以**批次到达**，到达间隔服从**指数分布**，均值为 10 分钟
- 每批次包含 **5–20 件**产品（离散均匀分布），产品类型按上述比例分配
- 各工位前的缓冲区容量为 **50 件**

## 五、换模时间

同一工位在不同产品之间切换时，需要换模时间：

| 切换情形 | 换模时间 |
|----------|----------|
| 同类型产品连续加工 | 0 分钟 |
| 不同类型产品切换 | 15 分钟（固定） |

## 六、设备故障

- 每台机床的平均故障间隔时间（MTBF）为 **100 小时**
- 平均修复时间（MTTR）为 **4 小时**（指数分布）
- 故障发生后，该工位的在制品**等待修复**

## 七、核心决策问题
1.	一周内，四种产品的产量分别是多少？总产量是多少？
2.	各工位的设备利用率分别是多少？哪个工位是瓶颈？
3.	各工位前的平均排队长度和平均等待时间是多少？
4.	换模时间对系统产能的影响有多大？
5.	如果在M4工位增加1台机床（共2台），系统产能提升多少？
6.	如果优化生产排序（减少产品切换次数），产能能提高多少？
7.	设备故障导致的产能损失是多少？
